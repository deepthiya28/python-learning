Qn 12. Armstrong Number Check Get a non-negative integer and check whether it is an Armstrong number. For example, 153 is an Armstrong
number because 1³ + 5³ + 3³ = 153.

```python
n = int(input("Enter a non-negative integer: "))
original = n
digits = len(str(n))
total = 0

while n > 0:
    digit = n % 10
    total += digit ** digits
    n //= 10

if total == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
n = int(input("Enter an integer: "))
original = n
temp = abs(n)
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp //= 10

if n >= 0 and reverse == original:
    print("Palindrome")
else:
    print("Not a palindrome")
```

Qn 16. Repeated Digit Sum, Get a positive integer and repeatedly add its digits until only one digit remains. Display each intermediate sum and the final single digit.

```python
n = int(input("Enter a positive integer: "))

while n >= 10:
    total = 0

    while n > 0:
        digit = n % 10
        total += digit
        n //= 10

    print("Intermediate sum =", total)
    n = total

print("Final single digit =", n)
```

Qn 19. Strong Number Check Get a positive integer and check whether it is a Strong number. A number is Strong if the sum of the factorials of its digits equals the original number. Example: 145 = 1! + 4! + 5!.

```python
n = int(input("Enter a positive integer: "))
original = n
total = 0

while n > 0:
    digit = n % 10
    factorial = 1

    for i in range(1, digit + 1):
        factorial *= i

    total += factorial
    n //= 10

if total == original:
    print("Strong number")
else:
    print("Not a Strong number")
```

Qn 3. Given a dictionary of student marks, write a program to calculate and print the total marks,
average marks, highest mark, and lowest mark using a loop.


```python
marks = {
    "John": 85,
    "Mary": 90,
    "David": 75,
    "Sarah": 95,
    "Peter": 80
}

total = 0
highest = list(marks.values())[0]
lowest = list(marks.values())[0]

for mark in marks.values():
    total = total + mark

    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

average = total / len(marks)

print("Total marks:", total)
print("Average marks:", average)
print("Highest mark:", highest)
print("Lowest mark:", lowest)
```
