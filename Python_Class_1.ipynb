```python
# Arithmetic Operators: +, -, *, /, %, **
a = 10      #compile time input - which is given in the program itself
b = int(input())   #run time input - which is given by the user
# Addition +
print(a+b)
# Subtraction -
print(a-b)
# Multiplication *
print(a*b)
# Division /
print(a/b)
# Floor Division //  (we don't want decimal points)
print(a//b)
# Modulo % (Gives the remainder of division, e.g., 8 % 3 = 2)
print(a%b)
# Exponentiation ** (power value)
print(a**b)
```

```python
#union, intersection, difference
a={1,2,3,4,5}
b={3,4,5,6,7}
print(a.union(b))
print(a.intersection(b))
print(b.difference(a))
```
